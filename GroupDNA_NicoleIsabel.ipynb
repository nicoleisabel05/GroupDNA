```python
# Assisted with AI
import numpy as np
from datetime import datetime, timedelta

file_path = 'hostel_bois.txt'
width = 60

months = ['January', 'February', 'March', 'April', 'May', 'June', 'July',
          'August', 'September', 'October', 'November', 'December']

# characters we remove from words (the apostrophe stays so "don't" survives)
punct = '.,!?;:"()[]{}<>-_*~`@#$%^&+=/\\|'

stop_words = ['i', 'is', 'the', 'a', 'an', 'and', 'or', 'to', 'of', 'in', 'on',
              'for', 'it', 'me', 'my', 'you', 'we', 'at', 'be', 'are', 'was',
              'so', 'that', 'this', 'with', 'not', 'do', 'have', 'but', 'if',
              'just', 'will', 'can', 'what', 'all', 'no', 'your', "i'm", "it's",
              'am', 'as', 'he', 'she', 'they', 'from', 'up', 'out', 'got',
              'how', 'about', 'today', 'his', 'her', 'which', 'who', 'why',
              'there', 'one', 'had', 'did', 'been', 'now', 'came', 'started',
              'entire', 'everyone', 'everything', 'anyone', 'telling', 'three',
              'used', 'hai']  # 'hai' = Hindi "is", same kind of filler as 'is'

caring_words = ['okay', 'safe', 'eat', 'sleep', 'take care', 'are you', 'please',
                'reminder', 'drink water', "don't forget"]
laugh_word = ['lol', 'lmao', 'haha', 'rofl', 'lmfao']
night_words = [23, 0, 1, 2, 3, 4]
```

```python
# 01.The chat parser
from datetime import datetime
with open (file_path, 'r', encoding = 'utf-8') as f:
  content =f.read()

content = content.replace('\u200e','').replace('\ufeff','')
lines = content.split('\n')

#AI assisted section
entries=[]
for line in lines:
  line=line.rstrip('\r')
  if line.strip() == '':
    continue

  startdate = (len(line) >= 10  and line[0:2].isdigit() and line[2] == '/' and line[3:5].isdigit() and line[5] == '/' and line[6:8].isdigit() and line[8] == ',')
  if startdate and '-' in line:
    parts = line.split('-', 1)
    entries.append([parts[0], parts[1]])
  elif len(entries)>0:
    entries[-1][1]=entries[-1][1] +'\n'+line.strip()

print("Total lines:", len(lines))
print("Recognized entries:", len(entries))
print("First 3 entries:", entries[:3])

message=[]
sy_count =0
med_count= 0
del_count=0
g_name='Hostel Bois 4ever'

for entry in entries:
  timestamp=entry[0].strip()
  body = entry[1]

  parts = body.split(':',1)
  if len(parts)<2:
    sy_count +=1
    continue
  sender = parts[0]
  text = parts[1].strip()

  if ':' not in body:
    sy_count +=1
    if 'created group' in body:
      g_name=body.split('"')[1]
    continue

  parts=body.split(':', 1)
  sender =parts[0].strip()
  text= parts[1].strip()

  if len(sender)>30:
    sy_count +=1
    continue

  if text =='<Media omitted>':
    kind='media'
    med_count +=1

  elif text in (
      'This message was deleted',
      'This message was deleted.',
      'You deleted this message'
      ):
    kind='deleted'
    del_count +=1

  else:
    kind='text'

  try:
    dt=datetime.strptime(timestamp,'%d/%m/%y, %H:%M')

  except ValueError as e:
    print("Timestamp error:", repr(timestamp), e)
    sy_count+=1
    continue
  message.append({
    'dt': dt,
    'date': timestamp[:8],
    'hour': dt.hour,
    'sender': sender,
    'text': text,
    'kind': kind
    })

print("Timestamp:", repr(timestamp))
print("Body:", repr(body))
print("Total entries detected:", len(entries))
print("Messages parsed:", len(message))
print("System messages skipped:", sy_count)
print("Deleted messages:", del_count)
print("Media messages:", med_count)


if len(message)>0:
  print("First parsed message:", message[0])
else:
  print("No messages found in the file. Is it a WhatsApp export?")
  raise SystemExit
```

```python
# Group Overview
counts={}
for m in message:
  if m['sender'] in counts:
    counts[m['sender']]+=1
  else:
    counts[m['sender']]=1
ranking=sorted([(counts[n],n)for n in counts], reverse=True)
names=[pair[1]for pair in ranking]
total_mes=len(message)

f_day=datetime.strptime(message[0]['date'], '%d/%m/%y')
l_day=datetime.strptime(message[-1]['date'], '%d/%m/%y')
t_day= (l_day- f_day).days+1

print(f"Messages successfully parsed {total_mes:,} messages from {len(names)} participants over {t_day} days, skipped{sy_count} system messages, {med_count} media-oitted, {del_count} deleted messages")
```

```python
# Activities during the days and hours
d_count={}
h_count={}

for m in message:
  if m['date'] in d_count:
    d_count[m['date']]+=1
  else:
    d_count[m['date']]=1
  if m['hour'] in h_count:
    h_count[m['hour']]+=1
  else:
    h_count[m['hour']] = 1

busy_day=''
busy_day_count=0
for d in d_count:
  if d_count[d] > busy_day_count:
    busy_day = d
    busy_day_count=d_count[d]

busy_h=0
busy_h_count=0
for h in h_count:
  if h_count[h] > busy_h_count:
    busy_h = h
    busy_h_count=h_count[h]

busy_day_dt=datetime.strptime(busy_day, '%d/%m/%y')

print(f"The busiest day is: {busy_day}")
print(f"The messages on the busiest day: {busy_day_count}")
print(f"The busiest hour {busy_h: 02d}:00-{(busy_h+1)% 24:02d}:00 ")
print(f"The messages on the busiest hour: {busy_h_count}")
```

```python
# Activity heatmap
row_of={}
for i in range(len(names)):
  row_of[names[i]]=i

heat= np.zeros((len(names), 24),dtype=int)
for m in message:
  heat[row_of[m['sender']], m['hour']] +=1
```

```python
# Person Statistics
group_word={}
person_word={}
stats={}
for n in names:
  person_word[n]={}
  stats[n]={'text': 0, 'words': 0, 'caring': 0, 'laughs':0, 'drama':0, 'questions': 0, 'media': 0, 'deleted': 0}

for m in message:
  s=stats[m['sender']]
  if m['kind'] == 'media':
    s['media'] +=1
    continue

  if m['kind'] =='deleted':
    s['deleted'] += 1
    continue

  text = m['text']
  s['text'] +=1
  s['words'] += len(text.split())

  if (len(text)>=3 and text.isupper()) or text.count("!")>=2:
    s['drama'] += 1

  if text.endswith('?'):
    s['questions']+=1

  clean=text.lower().replace('', "'")
  for ch in punct:
    clean=clean.replace(ch, ' ')
  words=[]
for w in clean.split():
  w=w.strip("'")
  if w!='':
    words.append(w)

padded = ' '+' '.join(words)+' '
for kw in caring_words:
  s['caring']+= padded.count(' '+ kw +'')

for w in words:
  if w in laugh_word:
    s['laughs']+=1
  if w in stop_words:
    continue
  if w in group_word:
    group_word[w]+=1
  else:
    group_word[w]=1
  if w in person_word[m['sender']]:
    person_word[m['sender']][w]+=1
  else:
    person_word[m['sender']][w]=1

top_group_words=sorted([(group_word[w],w) for w in group_word], reverse=True)[:10]

top_person_words={}
for n in names:
  top_person_words[n]=sorted([(person_word[n][w],w)for w in person_word[n]], reverse=True)[:5]
```

```python
# CODE TO CHECK IF THE PERSON STATISTICS IS WORKING
#AI ASSISTED

print("PARTICIPANT STATISTICS CHECK")

total_text = 0
total_media = 0
total_deleted = 0
total_words = 0

for n in names:
    s = stats[n]

    total_text += s['text']
    total_media += s['media']
    total_deleted += s['deleted']
    total_words += s['words']

    print(
        n,
        "| Text:", s['text'],
        "| Words:", s['words'],
        "| Media:", s['media'],
        "| Deleted:", s['deleted'],
        "| Caring:", s['caring'],
        "| Laughs:", s['laughs'],
        "| Drama:", s['drama'],
        "| Questions:", s['questions']
    )

print("\nVALIDATION")
print("Text messages:", total_text)
print("Media messages:", total_media)
print("Deleted messages:", total_deleted)
print("Total classified:", total_text + total_media + total_deleted)
print("Total parsed messages:", len(message))
print("Word total:", total_words)
print("Group word frequencies:", sum(group_word.values()))
print("Participant word frequencies:",
      sum(sum(person_word[n].values()) for n in names))

print("Message count matches:",
      total_text + total_media + total_deleted == len(message))

print("Word frequencies match:",
      sum(group_word.values()) ==
      sum(sum(person_word[n].values()) for n in names))

print("\nTOP 10 GROUP WORDS")
print(top_group_words)

print("\nTOP 5 WORDS PER PARTICIPANT")
for n in names:
    print(n, top_person_words[n])
```

```python
# Response Speed and Silent Streaks
gap_total={}
gap_number={}
for i in range(1, len(message)):
  before=message[i-1]
  now=message[i]
  if now['sender']!= before['sender']:
    seconds=(now['dt']-before['dt']).total_seconds()
    if now['sender'] in gap_total:
      gap_total[now['sender']]+=seconds
      gap_number[now['sender']]+=1
    else:
      gap_total[now['sender']]=seconds
      gap_number[now['sender']]=1

avg_gap={}
for n in gap_total:
  avg_gap[n]=gap_total[n]/gap_number[n]

act_days={}

for n in names:
  act_days[n]=[]

for m in message:
  act_days[m['sender']].append(m['date'])

long_streak={}
streak_start={}
streak_end={}
silent_days={}

for n in names:
  longest=0
  current=0
  current_start= None

  long_streak[n]= 0
  streak_start[n]= None
  streak_end[n]= None
  silent_days[n]= 0

  for i in range(t_day):
    day=f_day+ timedelta(days=i)
    day_text=f'{day.day:02d}/{day.month:02d}/{day.year% 100:02d}'
    if day_text in act_days[n]:
      current=0
    else:
      silent_days[n] +=1
      if current==0:
        current_start=day
        current+=1
        if current>longest:
          longest=current
          long_streak[n]=current
          streak_start[n]=current_start
          streak_end[n]=day
```

```python
# TO CHECK IF THE CODE ABOVE WORKS
# AI ASSISTED


print("RESPONSE SPEED")
for n in names:
    if n in avg_gap:
        print(n, "Average gap:", round(avg_gap[n] / 60, 2), "minutes")
    else:
        print(n, "No qualifying sender transitions")

print("\nSILENT DAYS AND LONGEST STREAK")
for n in names:
    print(
        n,
        "| Total silent days:", silent_days[n],
        "| Longest streak:", long_streak[n],
        "| Start:", streak_start[n],
        "| End:", streak_end[n]
    )

print("\nVALIDATION")
print("Participants checked:", len(silent_days))
print("All silent-day counts non-negative:",
      all(silent_days[n] >= 0 for n in names))
print("All streak lengths non-negative:",
      all(long_streak[n] >= 0 for n in names))
```

```python
# Personality Archetypes
brst_count = {}
for n in names:
    brst_count[n] = 0

prev_sender = ''

for m in message:
    if m['sender'] != prev_sender:
        brst_count[m['sender']] += 1
    prev_sender = m['sender']


rep_count = {}
shrt_count = {}
text_count = {}
pers_mess = {}
short_pct = {}

for n in names:
    rep_count[n] = 0
    shrt_count[n] = 0
    text_count[n] = 0
    pers_mess[n] = {}
    short_pct[n] = 0

for m in message:
    n = m['sender']

    if m['kind'] != 'text':
        continue

    text = m['text'].lower().strip()
    words = text.split()

    text_count[n] += 1

    if len(words) <= 2:
        shrt_count[n] += 1
        continue

    if text in pers_mess[n]:
        pers_mess[n][text] += 1
    else:
        pers_mess[n][text] = 1


for n in names:
    for text in pers_mess[n]:
        if pers_mess[n][text] >= 5:
            rep_count[n] += pers_mess[n][text] - 1


for n in names:
    if text_count[n] > 0:
        short_pct[n] = shrt_count[n] / text_count[n] * 100


night_hours = [23, 0, 1, 2, 3, 4]
all_pairs = []

for i in range(len(names)):
    n = names[i]
    s = stats[n]

    # For the Spammer
    if brst_count[n] > 0:
        avg_brst = counts[n] / brst_count[n]
    else:
        avg_brst = 0

    # For the Night Owl
    if heat[i].sum() > 0:
        night_pct = heat[i][night_hours].sum() / heat[i].sum() * 100
    else:
        night_pct = 0

    # For the Ghost
    if t_day > 0:
        ghost_pct = silent_days[n] / t_day * 100
    else:
        ghost_pct = 0

   # For the Group Mom
    if s['text'] > 0:
      caring_pct = s['caring'] / s['text'] * 100
    else:
      caring_pct = 0

    if s['text'] > 0:
        avg_words = s['words'] / s['text']
        drama_pct = s['drama'] / s['text'] * 100
        laugh_pct = s['laughs'] / s['text'] * 100
        question_pct = s['questions'] / s['text'] * 100
    else:
        avg_words = 0
        drama_pct = 0
        laugh_pct = 0
        question_pct = 0

    if counts[n] > 0:
        del_score = s['deleted'] / counts[n] * 100
        media_pct = s['media'] / counts[n] * 100
    else:
        del_score = 0
        media_pct = 0

    # For the Broken Record
    if text_count[n] > 0:
        rep_pct = rep_count[n] / text_count[n] * 100
    else:
        rep_pct = 0

    if rep_pct >= 50:
        broken_score = (rep_pct - 50) / 100
    else:
        broken_score = 0

    all_pairs.append((
        avg_brst / 3,
        n,
        'THE SPAMMER',
        f'{avg_brst:.1f} messages per burst'))
    all_pairs.append((
        caring_pct / 20,
        n,
        'THE GROUP MOM',
        f'Caring keywords score: {caring_pct:.1f}%' ))
    all_pairs.append((
        night_pct / 60,
        n,
        'THE NIGHT OWL',
        f'{night_pct:.1f}% messages between 23h-04h'))
    all_pairs.append((
        avg_words / 30,
        n,
        'THE STORY TELLER',
        f'{avg_words:.1f} words per message'))
    all_pairs.append((
        drama_pct / 30,
        n,
        'THE DRAMA QUEEN',
        f'{drama_pct:.1f}% shouty messages'))
    all_pairs.append((
        ghost_pct / 60,
        n,
        'THE GHOST',
        f'silent on {ghost_pct:.0f}% of days'))
    all_pairs.append((
        laugh_pct / 20,
        n,
        'THE COMEDIAN',
        f'{laugh_pct:.1f}% messages with a laugh word'))
    all_pairs.append((
        question_pct / 25,
        n,
        'THE QUESTION MASTER',
        f'{question_pct:.1f}% messages ending with ?'))
    all_pairs.append((
        del_score / 20,
        n,
        'THE TWEET AND DELETER',
        f'{del_score:.1f}% deletion notifications'))
    all_pairs.append((
        media_pct / 30,
        n,
        'THE MEDIA DROPPER',
        f'{media_pct:.1f}% media messages'))
    all_pairs.append((
        short_pct[n] / 60,
        n,
        'THE ONE-WORD WARRIOR',
        f'{short_pct[n]:.1f}% messages with 1-2 words'))
    all_pairs.append((
        broken_score,
        n,
        'THE BROKEN RECORD',
        f'{rep_count[n]} repeated longer messages '
        f'({rep_pct:.1f}% of text messages)'))


all_pairs = sorted(all_pairs, reverse=True)

archetype_of = {}
archetype_note = {}

for n in names:
    archetype_of[n] = []
    archetype_note[n] = []

for pair in all_pairs:
    score = pair[0]
    n = pair[1]
    label = pair[2]
    note = pair[3]

    if label not in archetype_of[n] and len(archetype_of[n]) < 2:
        archetype_of[n].append(label)
        archetype_note[n].append(note)


# ---------------- Display archetypes ----------------
print('GroupDNA Personality Archetypes')
print('=' * 40)

for n in names:
    print('\n' + n)

    for i in range(len(archetype_of[n])):
        print('-', archetype_of[n][i])
        print('   ', archetype_note[n][i])
```

```python

# Final Report
# to give each person their top 2 archetypes
all_pairs = sorted(all_pairs, reverse=True)
archetype_of = {}
archetype_note = {}

for n in names:
    archetype_of[n] = []
    archetype_note[n] = []

for pair in all_pairs:
    score = pair[0]
    n = pair[1]
    label = pair[2]
    note = pair[3]

    if len(archetype_of[n]) < 2:
        archetype_of[n].append(label)
        archetype_note[n].append(note)


line = '=' * width
print(line)
print(f'GroupDNA Report-"{g_name}"')
print(f'{t_day} days; {total_mes:,} messages; {len(names)} members')
print(line)

first_dt = message[0]['dt']
last_dt = message[-1]['dt']

print(
    f'Period: {first_dt.day:02d} {months[first_dt.month-1]} {first_dt.year}'
    f' to {last_dt.day:02d} {months[last_dt.month-1]} {last_dt.year}'
)
print(
    f'Busiest day: {busy_day_dt.day:02d} '
    f'{months[busy_day_dt.month-1]} {busy_day_dt.year} '
    f'({busy_day_count} messages)'
)
print(
    f'Busiest Hour: {busy_h:02d}:00-{busy_h+1:02d}:00 '
    f'(avg {busy_h_count/t_day:.0f} messages per day)'
)
print(f'Skipped: {sy_count} system, {med_count} media, {del_count} deleted')


# --------- Messages per person ---------
print()
print('Messages Per Person')

top_count = max(counts.values())

for n in names:
    if counts[n] == 0 or top_count == 0:
        bar = '.'
    else:
        bar = '█' * max(1, round(counts[n] / top_count * 20))

    print(f'{n:<8} {bar} {counts[n]}')


# --------- Heatmap ---------
print()
print('Activity Heat Map')
header = 'Name     '
for h in range(0, 24, 3):
    header += f'{h:02d} '

print(header)

for i in range(len(names)):
    row_max = heat[i].max()
    cells = ''
    for h in range(24):
        if row_max == 0:
            ratio = 0
        else:
            ratio = heat[i][h] / row_max
        if ratio < 0.25:
            cells += '.  '
        elif ratio < 0.5:
            cells += '░  '
        elif ratio < 0.75:
            cells += '▒  '
        else:
            cells += '█  '

    tag = ''
    if 'THE NIGHT OWL' in archetype_of[names[i]]:
        tag = '<- NIGHT OWL'

    print(f'{names[i]:<8} {cells}{tag}')


# --------- Top group words ---------
print()
print("The Group's Favourite Words")

if len(top_group_words) > 0:
    biggest = top_group_words[0][0]
    for pair in top_group_words:
        if biggest > 0:
            bar_length = max(1, round(pair[0] / biggest * 20))
        else:
            bar_length = 0
        bar = '█' * bar_length
        print(f' {pair[1]:<10} {bar:<20} {pair[0]}')
else:
    print('No group word data available.')


# --------- Top words per person ---------
print()
print('Top Words Per Person')

for n in names:
    word_list = ''
    for pair in top_person_words[n]:
        word_list += pair[1] + ','
    print(f'{n:<8} {word_list[:-1]}')

# --------- Response patterns ---------
print()
print('Response Patterns')

if len(avg_gap) > 0:
    pairs = sorted([(avg_gap[n], n) for n in avg_gap])
    for label, pair in [('Fastest', pairs[0]), ('Slowest', pairs[-1])]:
        if pair[0] < 3600:
            nice = f'{pair[0]/60:.1f} minutes'
        else:
            nice = f'{pair[0]/3600:.1f} hours'
        print(f'{label} replier: {pair[1]} (avg {nice})')
else:
    print('Not enough back-and-forth to measure.')


# --------- Silent streaks ---------
print()
print('Longest Silent Streaks')

streak_pairs = sorted(
    [(long_streak[n], n) for n in names],
    reverse=True
)

for streak, n in streak_pairs[:5]:
    print(f' {n:<8} {streak} days')


# --------- Media and deleted messages ---------
print()
print('Media and Deleted Messages')

for n in names:
    print(
        f' {n:<8} media: {stats[n]["media"]:<3} '
        f'deleted: {stats[n]["deleted"]}'
    )

all_pairs=sorted(all_pairs, reverse=True)
archetype_of={}
archetype_note={}

for n in names:
  archetype_of[n]=[]
  archetype_note[n]=[]
used_archetypes=[]

for pair in all_pairs:
  score=pair[0]
  n=pair[1]
  label=pair[2]
  note=pair[3]

  if label not in used_archetypes and len(archetype_of[n])<2:
    archetype_of[n].append(label)
    archetype_note[n].append(note)
    used_archetypes.append(label)


# --------- Personality archetypes ---------
print()
print('GroupDNA Personality Archetypes')
print('=' * 40)

for n in names:
    print('\n' + n)

    for i in range(len(archetype_of[n])):
        print('-', archetype_of[n][i])
        print('   ', archetype_note[n][i])

print()
print(line)
```
